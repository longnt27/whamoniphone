# Export WHAM with image features

This notebook delegates to `utils/wham_coreml.py`, the same implementation used by the CPU tests. It replaces the old inline export that bypassed the integrator.

Run inside this repository with an existing upstream WHAM checkout and trusted checkpoint. No GPU or SMPL body data is needed for this neural-network export. See `docs/image-features.md` for dependencies and migration.

Both WHAM and FastViT packages must be re-exported. This notebook only exports WHAM.


```python
from pathlib import Path
import subprocess
import sys

REPO = next((p for p in (Path.cwd(), Path.cwd().parent)
             if (p / "utils/wham_coreml.py").is_file()), None)
if REPO is None:
    raise RuntimeError("Open this notebook from the whamoniphone checkout.")

# Edit these paths for your machine.
WHAM_ROOT = Path.home() / "WHAM"
CHECKPOINT = WHAM_ROOT / "checkpoints/wham_vit_bedlam_w_3dpw.pth.tar"
OUTPUT = REPO / "exports"
TRUST_CHECKPOINT = False  # Enable only after verifying the checkpoint provenance.
```

```python
subprocess.run([sys.executable, "-m", "unittest", "discover", "-s", "tests", "-v"],
               cwd=REPO, check=True)
```

```python
command = [sys.executable, str(REPO / "utils/wham_coreml.py"),
           "--wham-root", str(WHAM_ROOT), "--checkpoint", str(CHECKPOINT),
           "--out", str(OUTPUT)]
if TRUST_CHECKPOINT:
    command.append("--trust-checkpoint")
if sys.platform == "darwin":
    command.append("--verify-coreml")
subprocess.run(command, cwd=REPO, check=True)
```
